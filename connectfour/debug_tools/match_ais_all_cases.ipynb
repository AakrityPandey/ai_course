# Connect Four: two configurable AIs

Choose each player's Python file in the Settings cell. The notebook plays every starting board without opening the game window. A timed-out or invalid move is wasted, as in the simulator. Games stop after 12 consecutive wasted turns or at a safety cap.


```python
from pathlib import Path
import sys
from IPython.display import Markdown, display

# Locate the project whether the notebook kernel starts in debug_tools,
# starter_code, or a parent folder.
search_roots = (Path.cwd(), *Path.cwd().parents)
candidates = [
    candidate
    for root in search_roots
    for candidate in (root / "connectfour" / "debug_tools", root / "debug_tools")
    if (candidate / "batch_match_ais.py").is_file()
]
if not candidates:
    raise FileNotFoundError("Could not find connectfour/debug_tools/batch_match_ais.py")
DEBUG_DIR = candidates[0].resolve()
STARTER_DIR = DEBUG_DIR.parent / "starter_code"
sys.path.insert(0, str(STARTER_DIR))
sys.path.insert(0, str(DEBUG_DIR))
import importlib
import batch_match_ais
importlib.reload(batch_match_ais)  # Pick up changes without restarting the notebook kernel.
play_case = batch_match_ais.play_case

print(f"Project: {DEBUG_DIR.parent}")
print(f"My AI: {STARTER_DIR / 'connect4/players/ai.py'}")
print(f"Tanmay AI: {STARTER_DIR / 'connect4/players/ai_tanmay_noob.py'}")
```

## Settings

Set `PLAYER_1_FILE` and `PLAYER_2_FILE` to filenames in `starter_code/connect4/players`. Swap the two values to swap sides. `TIME_LIMIT` is the number of seconds allowed per move.


```python
PLAYER_1_FILE = "ai.py"
PLAYER_2_FILE = "ai_tanmay_noob.py"
TIME_LIMIT = 10
  # seconds per move
CASE_FILES = sorted((STARTER_DIR / "connect4" / "initial_states").glob("case*.txt"))
print(f"Player 1: {PLAYER_1_FILE}; Player 2: {PLAYER_2_FILE}")
print("Cases:", ", ".join(path.name for path in CASE_FILES))
```

## Run all cases

This cell can take several minutes if you increase `TIME_LIMIT`. It prints each result as soon as that case finishes.

```python
results = []
for case_file in CASE_FILES:
    result, player1_score, player2_score = play_case(
        case_file, TIME_LIMIT, return_scores=True,
        player1_file=PLAYER_1_FILE, player2_file=PLAYER_2_FILE,
    )
    results.append({
        "case": case_file.name,
        "player1_score": player1_score,
        "player2_score": player2_score,
        "result": result,
    })
    print(f"{case_file.name}: {PLAYER_1_FILE} = {player1_score}, {PLAYER_2_FILE} = {player2_score}; {result}", flush=True)
```

## Results table

Scores are final when a game finishes. If a case stalls or reaches the safety cap, the scores shown are those reached at that point.


```python
rows = ["| Board | Player 1 (Player 1) | Player 2 (Player 2) | Result |", "|---|---:|---:|---|"]
rows.extend(
    f"| {item['case']} | {item['player1_score']} | {item['player2_score']} | {item['result']} |"
    for item in results
)
display(Markdown("\n".join(rows)))
```
