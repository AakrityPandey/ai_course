# Chess — responsive desktop board

Select the shared `ai_course/.venv` Python kernel and run the two code cells below.
The Pygame window runs on the main process; AI moves are calculated in a separate process so the board stays responsive.
The status bar shows **thinking...** during search. Close the board to cancel the search and finish the cell.

Start with `SEARCH_DEPTH = 3`. Higher depths can take much longer; a responsive window does not make the search faster.

Minimax branch score printing is disabled.


```python
from pathlib import Path
import importlib
import random
import sys

NOTEBOOK_DIR = Path.cwd()
if not (NOTEBOOK_DIR / "chess_engine").is_dir():
    candidates = (
        NOTEBOOK_DIR / "ai_course" / "chess",
        NOTEBOOK_DIR / "my_ml" / "chess",
        NOTEBOOK_DIR.parent / "my_ml" / "chess",
        NOTEBOOK_DIR / "chess",
        NOTEBOOK_DIR.parent / "chess",
    )
    NOTEBOOK_DIR = next(
        (candidate for candidate in candidates if (candidate / "chess_engine").is_dir()),
        candidates[0],
    )
sys.path.insert(0, str(NOTEBOOK_DIR))

import chess_engine
for module_name in ("state", "rules", "ai", "players"):
    importlib.reload(importlib.import_module(f"chess_engine.{module_name}"))
importlib.reload(chess_engine)

import pygame_ui
importlib.reload(pygame_ui)

from chess_engine import BoardState, Player
from pygame_ui import PygameChessUI
```

```python
def play_game(player1_mode="manual", player2_mode="minimax", depth=3):
    """Start Pygame chess and randomly assign the players to colors."""
    state = BoardState.initial(turn=1)
    player1 = ("Player 1", Player(player1_mode, depth=depth))
    player2 = ("Player 2", Player(player2_mode, depth=depth))
    white_player, black_player = random.sample([player1, player2], 2)
    players = {1: white_player, -1: black_player}

    print("{} is White; {} is Black.".format(white_player[0], black_player[0]))
    return PygameChessUI(state, players)


# Game settings: change these values, then rerun this cell.
PLAYER_1_MODE = "manual"  # "manual", "random", or "minimax"
PLAYER_2_MODE = "minimax"  # "manual", "random", or "minimax"
SEARCH_DEPTH = 3  # Used by players in minimax mode

# Player 1 and Player 2 are randomly assigned White and Black.
game = play_game(
    player1_mode=PLAYER_1_MODE,
    player2_mode=PLAYER_2_MODE,
    depth=SEARCH_DEPTH,
)
game.run()
```
